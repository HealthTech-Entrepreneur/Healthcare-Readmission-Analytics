```python
# 1. Count total duplicate rows
duplicate_count = df.duplicated().sum()
print(f"Total duplicate rows: {duplicate_count}")

# 2. View the actual duplicate rows (if any exist)
if duplicate_count > 0:
    print("\n--- SAMPLE DUPLICATE ROWS ---")
    display(df[df.duplicated(keep=False)].sort_values(by=list(df.columns[:2])).head(10))
else:
    print("No duplicate rows found!")
```
