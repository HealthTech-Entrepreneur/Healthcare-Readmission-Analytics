```python
import pandas as pd

import numpy as np
```

```python
df = pd.read_csv("../Files/processed/clean_healthcare_readmissions.csv")
```

```python
df.head(5)
```

Define the key identifier columns to place at the beginning

```python
key_columns = ['Record_ID', 'Encounter_ID', 'Patient_ID', 'Provider_ID']
```

Getting the remaining columns in their existing order

```python

other_columns = [col for col in df.columns if col not in key_columns]
```

Reorder DataFrame columns

```python
df_reordered = df[key_columns + other_columns]
```

Saving the reordered DataFrame back to the CSV file

```python
file_path = "../Files/processed/clean_healthcare_readmissions.csv"
```

```python
df_reordered.to_csv(file_path, index=False)
```
